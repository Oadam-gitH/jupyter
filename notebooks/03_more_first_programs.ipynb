_Suppose you purchase a 10 inch pizza. How many square *centimetres* of pizza do you have? Write a program below to find out. Remember that 10 inches is the diameter of the pizza._

```python
def calculate_pizza_area(diameter_inches):
    inches_to_cm = 2.54
    radius_inches = diameter_inches / 2
    radius_cm = radius_inches * inches_to_cm
    area_cm2 = math.pi * (radius_cm ** 2)
    return area_cm2

diameter = 10
area = calculate_pizza_area(diameter)

print(f"A {diameter}-inch pizza has a radius of {diameter/2 * 2.54:.1f} cm.")
print(f"Total area: {area:.2f} square centimeters.")
```

_A 10 inch pizza costs £10.99. A 12 inch pizza costs £13.99. Which is the best buy in terms of pizza area for your pound? Write a program
that prompts for the size and cost, and prints the cost per square cm._

_Hint: It is probably easier to work in pence per square cm. Even then, the answers will be small numbers._

```python
def calculate_pizza_value():
    diameter = float(input("enter the pizza diameter in inches: "))
    cost_pounds = float(input("enter the cost of the pizza (£): "))
    radius_cm = (diameter / 2) * 2.54
    area_cm2 = math.pi * (radius_cm ** 2)
    cost_pence = cost_pounds * 100
    pence_per_cm2 = cost_pence / area_cm2

    print(f"\nresults for a {diameter}inch pizza:")
    print(f"  total Area: {area_cm2:.2f} square cm")
    print(f"  value Rating: {pence_per_cm2:.4f}p per square cm")

if __name__ == "__main__":
    calculate_pizza_value()
```
